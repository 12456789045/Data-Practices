```python

import numpy as np
import pandas as pd

np.random.seed(42)

# Four underlying customer profiles
profiles = {
    "Budget": {
        "income": 30000, "spending": 25,
        "frequency": 3, "order": 700,
        "engagement": 25, "loyalty": 2
    },
    "Premium": {
        "income": 140000, "spending": 85,
        "frequency": 15, "order": 6500,
        "engagement": 90, "loyalty": 9
    },
    "Regular": {
        "income": 70000, "spending": 55,
        "frequency": 9, "order": 2200,
        "engagement": 60, "loyalty": 6
    },
    "Occasional": {
        "income": 50000, "spending": 35,
        "frequency": 2, "order": 1200,
        "engagement": 20, "loyalty": 3
    }
}

rows = []

for i in range(600):
    group = np.random.choice(
        list(profiles.keys()),
        p=[0.25, 0.20, 0.35, 0.20]
    )

    p = profiles[group]

    rows.append({
        "customer_id": f"C{i+1:04d}",
        "income": max(5000, np.random.normal(p["income"], 12000)),
        "spending_score": np.clip(
            np.random.normal(p["spending"], 12), 1, 100
        ),
        "purchase_frequency": max(
            0, round(np.random.normal(p["frequency"], 2))
        ),
        "average_order_value": max(
            100, np.random.normal(p["order"], p["order"] * 0.25)
        ),
        "website_engagement": np.clip(
            np.random.normal(p["engagement"], 15), 0, 100
        ),
        "loyalty_years": max(
            0, np.random.normal(p["loyalty"], 1.5)
        ),
        "age": np.clip(np.random.normal(35, 11), 18, 75)
    })

df = pd.DataFrame(rows)

# Remove the underlying labels: clustering is unsupervised
# Add realistic missing values
for col in [
    "income", "spending_score", "purchase_frequency",
    "average_order_value", "website_engagement",
    "loyalty_years", "age"
]:
    mask = np.random.rand(len(df)) < 0.06
    df.loc[mask, col] = np.nan

df.to_csv("hierarchical_customers.csv", index=False)

print(df.head())
print("\nMissing values:\n", df.isnull().sum())
print("\nDataset shape:", df.shape)
```
